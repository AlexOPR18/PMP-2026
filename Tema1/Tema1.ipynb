```python
import csv
import numpy as np
import pandas as pd


def alege_studenti(nume_fisier, numar_studenti_doriti):
    studenti = []

    try:
        with open(nume_fisier, mode='r', encoding='utf-8-sig') as fisier:
            prima_linie = fisier.readline()
            separator = ';' if ';' in prima_linie else ','
            fisier.seek(0)

            cititor = csv.reader(fisier, delimiter=separator)

            for rand in cititor:
                if not rand:
                    continue

                if len(rand) > 1 and rand[1].strip() != "":
                    nume = rand[1].strip()
                else:
                    nume = rand[0].strip()

                if nume.lower() not in ["", "nume", "name"]:
                    studenti.append(nume)

    except FileNotFoundError:
        return f"Eroare! N-am gasit fisierul: {nume_fisier}"

    if numar_studenti_doriti > len(studenti):
        return f"Eroare! Ai cerut {numar_studenti_doriti} studenti, dar fisierul are doar {len(studenti)}."

    studenti_alesi = np.random.choice(studenti, size=numar_studenti_doriti, replace=False)

    return studenti_alesi


print("Studentii care vor prezenta tema sunt:")
print(alege_studenti('PycharmProjects/WelcomeScreen/studenti.csv', 3))

pd.read_csv('PycharmProjects/WelcomeScreen/studenti.csv')
```

```python
import numpy as np
import matplotlib.pyplot as plt

def joaca(probabilitate_stema=0.5):
    N=0
    S=0
    while True:
        N+=1
        stema=np.random.rand() < probabilitate_stema

        if stema:
            zar = np.random.randint(1,7)
            S+= (zar - 3)
            break
        else:
            S-=0.5

    return N, S

def sim_and_plot(prob_stema, num_simulari=10000):
    sume=[]
    for _ in range(num_simulari):
        _, suma = joaca(prob_stema)
        sume.append(suma)

    media_S = np.mean(sume)
    print(f"Pentru p={prob_stema}: Media sumei S este aprox. {media_S: .3f}$")

    plt.hist(sume, bins=30, alpha=0.7,edgecolor='black')
    plt.title(f'Distributia sumei S pentru p={prob_stema}')
    plt.xlabel('Suma ($)')
    plt.ylabel('Frecventa')
    plt.axvline(media_S, color='red', linestyle='dashed', linewidth=2, label=f'Media: {media_S:.2f}')
    plt.legend()
    plt.show()

sim_and_plot(0.5)
sim_and_plot(0.3)
sim_and_plot(0.7)
```

```python
import numpy as np
import arviz as az
import matplotlib.pyplot as plt

viteze_frizeri = [3, 6, 4]
probabilitati_preluare = [3/13, 6/13, 4/13]
numar_clienti = 10000

frizeri_alocati = np.random.choice([0, 1, 2], p=probabilitati_preluare, size=numar_clienti)

timpi_servire_X = np.zeros(numar_clienti)

for i in range(numar_clienti):
    frizer_curent = frizeri_alocati[i]
    lambda_curent = viteze_frizeri[frizer_curent]
    timpi_servire_X[i] = np.random.exponential(scale=1/lambda_curent)

media_X = np.mean(timpi_servire_X)
deviatia_X = np.std(timpi_servire_X)

print(f"Timpul mediu estimat de servire (X) este: {media_X:.4f} ore")
print(f"Deviatia standard estimata a lui X este: {deviatia_X:.4f} ore")

az.plot_kde(timpi_servire_X)
plt.title("Densitatea aproximativa a distributiei lui X (Timp de servire)")
plt.xlabel("Timp de servire (ore)")
plt.ylabel("Densitate")
plt.show()
```
